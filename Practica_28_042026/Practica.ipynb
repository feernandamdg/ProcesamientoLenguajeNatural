## CORPUS Y PRE-PROCESAMIENTO 

```python
corpus = [
    "Caminando corriendo montando",
    "Esta película fue terrible y aburrida",
    "Los actores actuaron muy bien ",
    "No me gustaron las tramas fueron malas",
    "Excelente dirección y guion brillante"
] # Corpus de 1800 - 2000 palabras 
```

```python
# pre - procesamiento del corpus 
# mandar a llamar a la función de preprocesamiento
```

```python
import numpy as np

X = np.array([
    [1, 0.3, 0.5, 0.7, 0.5, 0.3],
    [0.7, 0.5, 0.3, 0.4, 0.3, 0.6],
    [0.4, 0.3, 0.1, 0.1, 0.2, 0.3],
    [0.1, 0.2, 0.3, 0.4, 0.3, 0.6]
])

Y = np.array([
    [0.4, 0.3, 0.6],
    [0.1, 0.2, 0.3],
    [0.4, 0.1, 0.4],
    [0.2, 0.1, 0.2]
])
```

## Red neuronal simple

```python
# pesos iniciales
np.random.seed(0)
W = np.random.rand(6, 3)

# entrenamiento
lr = 0.01

for epoch in range(1000):
    # forward
    Y_pred = np.dot(X, W)
    
    # error
    error = Y - Y_pred
    
    # gradiente
    grad = np.dot(X.T, error)
    
    # actualizar pesos
    W += lr * grad

print("Pesos entrenados:")
print(W)
```

```python
# ejemplo: Hola + Mundo
entrada = np.array([[1, 0.3, 0.5, 0.7, 0.5, 0.3]])

pred = np.dot(entrada, W)

print("Predicción vectorial:")
print(pred)
```

### Clasificar la predicción

```python
def coseno(a, b):
    return np.dot(a, b) / (np.linalg.norm(a) * np.linalg.norm(b))

# diccionario palabra → vector
vocab = {
    "ESCOM": np.array([0.9, 0.3, 0.6]),
    "es": np.array([0.1, 0.2, 0.3]),
    "lugar": np.array([0.4, 0.1, 0.4]),
    "donde": np.array([0.2, 0.1, 0.2])
}

mejor = ""
mejor_sim = -1

for palabra in vocab:
    sim = coseno(pred[0], vocab[palabra])
    
    if sim > mejor_sim:
        mejor_sim = sim
        mejor = palabra

print("Palabra predicha:", mejor)
```
