```python
def distancia(p1,p2):
    dist = 0
    todos = set(p1.keys()) | (set(p2.keys()))

    for k in todos:
        if k in p1: 
            v1 = p1[k]
        else:
            v1 = 0

        if k in p2:
            v2 = p2[k]
        else:
            v2 = 0
        if v1 > v2:
            dist += (v1 - v2)
        else:            
            dist += (v2 - v1)
    return dist
```
