```python
# Corpus como lista (evitamos .keys())
corpus = [
    ("doc1", "Einstein persona importante ambito matemáticas utilizó matemáticas como herramienta poderosa para remodelar física"),
    ("doc2", "Ciencias naturales estudian mundo físico natural abarcando organismos vivos procesos geológicos cósmicos"),
    ("doc3", "física ciencia natural estudia componentes fundamentales universo. materia energía espacio tiempo interacciones")
]

n_docs = len(corpus)

# TOKENIZADOR MANUAL
def tokenizar(texto):
    max_len = len(texto)
    palabras = [""] * max_len  # tamaño máximo posible
    palabra_actual = ""
    count = 0
    
    i = 0
    while i < len(texto):
        char = texto[i]
        
        if char != " ":
            palabra_actual = palabra_actual + char
        else:
            if palabra_actual != "":
                palabras[count] = palabra_actual
                count += 1
                palabra_actual = ""
        
        i += 1
    
    # última palabra
    if palabra_actual != "":
        palabras[count] = palabra_actual
        count += 1
    
    # recortar lista real
    resultado = [""] * count
    for i in range(count):
        resultado[i] = palabras[i]
    
    return resultado


# CONSTRUIR VOCABULARIO
max_words = 0

i = 0
while i < n_docs:
    tokens = tokenizar(corpus[i][1])
    max_words += len(tokens)
    i += 1

vocab = [""] * max_words
vocab_size = 0

i = 0
while i < n_docs:
    tokens = tokenizar(corpus[i][1])
    
    t = 0
    while t < len(tokens):
        palabra = tokens[t]
        
        existe = False
        k = 0
        while k < vocab_size:
            if vocab[k] == palabra:
                existe = True
                break
            k += 1
        
        if not existe:
            vocab[vocab_size] = palabra
            vocab_size += 1
        
        t += 1
    
    i += 1

# recortar vocabulario
vocab_final = [""] * vocab_size
i = 0
while i < vocab_size:
    vocab_final[i] = vocab[i]
    i += 1


# CREAR MATRIZ -------------------------------------------------------
matrix = []
i = 0
while i < n_docs:       # n_docs = len(corpus)
    fila = [0] * vocab_size
    matrix += [fila]   # evitamos append directo
    i += 1


# LLENAR MATRIZ
i = 0
while i < n_docs:
    tokens = tokenizar(corpus[i][1])
    
    t = 0
    while t < len(tokens):
        palabra = tokens[t]
        
        j = 0
        while j < vocab_size:
            if vocab_final[j] == palabra:
                matrix[i][j] += 1
                break
            j += 1
        
        t += 1
    
    i += 1

# OUTPUT
print("Vocabulario:")
print(vocab_final)

print("\nMatriz Bag of Words:")
i = 0
while i < n_docs:
    print(corpus[i][0], matrix[i])
    i += 1
```
