# N-Gramas
-> Son, en particular una secuencia contigua de "n" elementos (por ejemplo palabras * caracteres) extraídos de un corpus.

En el caso de 
* N=1 -> Unigrama
* N=2 -> Bigramas
* N=3 -> Trigramas

Ejemplo: 
texto: Hoy hace un calor infernal. No se asoleen.

Unigrama = [Hoy, hace, un, ...]
Bigrama = [Hoy hace, un calor, ...]
Trigrama = [Hoy hace un, hace un calor, ... ]

En general los N-gramas capturan dependencias locales y capturan la probabilidad de que una palabra aparezca a lado de otra.

Aplicaciones: 
- Predecir la sig. palabra
- Detección de idioma
- Conector ortografico

En el caso de N-gramas por letras.
Bigrama = [ho, oy, ha, ac, ce, ...] -> Hoy Hace 

Los N-gramas apoyan en capturar la morfologia linguistica y por tanto hacer detecciones de idiomas o correciones ortograficas.

Morfolgía -> Morfología  



Metodología que nos permite capturar contexto, se parece a la matriz de co-ocurrencia (donde se formaban parejas)

Diferencia: Matriz de coocurrencias solo es para parejas

N-Gramas:estos se pueden utilizar para palabras, oraciones y letras

Bigramas - asociarles un elemento de salida, sirve para hacer predicción de palabras
+ Cuando se combinan Bigramas, trigramas... permite capturar tamaño de dif relaciones semanticas, se captura 1,2,3,... localidades, ventaja de la generalización (va generalizando el contexto).

Bi, Tri, ya no generaliza, es muy local, generalmente se combinan y se meten a un algoritmo. 

Modelado de textos mediante localidades

A medida que crecen los N-gramas también crece el poder computacional, + vectores, vectores + grandes, mayor N en el n-grama se generaliza más

En la práctica: Bi, Tri y cuatri.

Ventaja: Se puede hacer por oraciones y  letras
letras: captar la morfologia del vocabulario, puedo tener N-gramas por letras.
